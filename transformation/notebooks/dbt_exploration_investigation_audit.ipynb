```python
query = """
WITH jsearch AS (
    SELECT
        RAW_PAYLOAD:job_id::string                           as job_id,
        'jsearch'                                            as source,
        RAW_PAYLOAD:job_title::string                        as job_title,
        RAW_PAYLOAD:employer_name::string                    as company_name,
        RAW_PAYLOAD:job_city::string                         as city,
        RAW_PAYLOAD:job_state::string                        as state,
        RAW_PAYLOAD:job_is_remote::boolean                   as is_remote,
        RAW_PAYLOAD:job_posted_at_datetime_utc::timestamp_tz as posted_at,
        RAW_PAYLOAD:job_min_salary::float                    as min_salary,
        RAW_PAYLOAD:job_max_salary::float                    as max_salary,
        RAW_PAYLOAD:job_salary_period::string                as salary_period,
        RAW_PAYLOAD:job_employment_type::string              as employment_type,
        RAW_PAYLOAD:job_description::string                  as description,
        RAW_PAYLOAD:job_apply_link::string                   as apply_url,
        INGESTED_AT
    FROM RAW.JSEARCH.SRC_POSTINGS
),

theirstack AS (
    SELECT
        RAW_PAYLOAD:id::string                               as job_id,
        'theirstack'                                         as source,
        RAW_PAYLOAD:job_title::string                        as job_title,
        RAW_PAYLOAD:company::string                          as company_name,
        NULL                                                 as city,
        RAW_PAYLOAD:state_code::string                       as state,
        RAW_PAYLOAD:remote::boolean                          as is_remote,
        RAW_PAYLOAD:date_posted::timestamp_tz                as posted_at,
        RAW_PAYLOAD:min_annual_salary_usd::float             as min_salary,
        RAW_PAYLOAD:max_annual_salary_usd::float             as max_salary,
        NULL                                                 as salary_period,
        RAW_PAYLOAD:employment_statuses[0]::string           as employment_type,
        RAW_PAYLOAD:description::string                      as description,
        RAW_PAYLOAD:source_url::string                       as apply_url,
        INGESTED_AT
    FROM RAW.THEIRSTACK.SRC_POSTINGS
),

builtin AS (
    SELECT
        RAW_PAYLOAD:identifier:value::string                       as job_id,
        'builtin'                                                  as source,
        RAW_PAYLOAD:title::string                                  as job_title,
        RAW_PAYLOAD:hiringOrganization:name::string                as company_name,
        RAW_PAYLOAD:jobLocation:address:addressLocality::string    as city,
        RAW_PAYLOAD:jobLocation:address:addressRegion::string      as state,
        CASE
            WHEN RAW_PAYLOAD:jobLocationType::string = 'TELECOMMUTE'
            THEN TRUE ELSE FALSE
        END                                                        as is_remote,
        RAW_PAYLOAD:datePosted::timestamp_tz                       as posted_at,
        CASE
            WHEN RAW_PAYLOAD:baseSalary:value:unitText::string = 'YEAR'
            THEN RAW_PAYLOAD:baseSalary:value:minValue::float
            ELSE NULL
        END                                                        as min_salary,
        CASE
            WHEN RAW_PAYLOAD:baseSalary:value:unitText::string = 'YEAR'
            THEN RAW_PAYLOAD:baseSalary:value:maxValue::float
            ELSE NULL
        END                                                        as max_salary,
        RAW_PAYLOAD:baseSalary:value:unitText::string              as salary_period,
        RAW_PAYLOAD:employmentType::string                         as employment_type,
        RAW_PAYLOAD:description::string                            as description,
        RAW_PAYLOAD:source_url::string                             as apply_url,
        INGESTED_AT
    FROM RAW.BUILTIN.SRC_POSTINGS
)

SELECT * FROM jsearch
UNION ALL
SELECT * FROM theirstack
UNION ALL
SELECT * FROM builtin
ORDER BY posted_at DESC
"""

df = pd.read_sql(query, conn)

print(f"Total rows: {len(df)}")
print(f"By source:")
print(df["SOURCE"].value_counts().to_string())
print()
print(f"Null counts per column:")
print(df.isnull().sum().to_string())
print()
print(f"Sample rows across all sources:")
print(df[["SOURCE", "JOB_TITLE", "COMPANY_NAME", "CITY", 
          "POSTED_AT", "MIN_SALARY", "IS_REMOTE"]].to_string())
```

```python
query = """
WITH all_titles AS (
    SELECT
        'jsearch' as source,
        RAW_PAYLOAD:job_title::string as job_title
    FROM RAW.JSEARCH.SRC_POSTINGS
    
    UNION ALL
    
    SELECT
        'theirstack' as source,
        RAW_PAYLOAD:job_title::string as job_title
    FROM RAW.THEIRSTACK.SRC_POSTINGS
    
    UNION ALL
    
    SELECT
        'builtin' as source,
        RAW_PAYLOAD:title::string as job_title
    FROM RAW.BUILTIN.SRC_POSTINGS
)

SELECT
    source,
    COUNT(*) as total_jobs,
    SUM(CASE WHEN REGEXP_LIKE(job_title, '.*\\bjunior\\b.*', 'i')     THEN 1 ELSE 0 END) as has_junior,
    SUM(CASE WHEN REGEXP_LIKE(job_title, '.*\\bentry\\b.*', 'i')      THEN 1 ELSE 0 END) as has_entry,
    SUM(CASE WHEN REGEXP_LIKE(job_title, '.*\\bassociate\\b.*', 'i')  THEN 1 ELSE 0 END) as has_associate,
    SUM(CASE WHEN REGEXP_LIKE(job_title, '.*\\bmid\\b.*', 'i')        THEN 1 ELSE 0 END) as has_mid,
    SUM(CASE WHEN REGEXP_LIKE(job_title, '.*\\bsenior\\b.*', 'i')     THEN 1 ELSE 0 END) as has_senior,
    SUM(CASE WHEN REGEXP_LIKE(job_title, '.*\\bsr\\b.*', 'i')         THEN 1 ELSE 0 END) as has_sr,
    SUM(CASE WHEN REGEXP_LIKE(job_title, '.*\\blead\\b.*', 'i')       THEN 1 ELSE 0 END) as has_lead,
    SUM(CASE WHEN REGEXP_LIKE(job_title, '.*\\bprincipal\\b.*', 'i')  THEN 1 ELSE 0 END) as has_principal,
    SUM(CASE WHEN REGEXP_LIKE(job_title, '.*\\bmanager\\b.*', 'i')    THEN 1 ELSE 0 END) as has_manager,
    SUM(CASE WHEN REGEXP_LIKE(job_title, '.*\\bdirector\\b.*', 'i')   THEN 1 ELSE 0 END) as has_director,
    SUM(CASE WHEN REGEXP_LIKE(job_title, '.*(\\bvp\\b|vice president).*', 'i') THEN 1 ELSE 0 END) as has_vp,
    SUM(CASE WHEN REGEXP_LIKE(job_title, '.*(\\bii\\b|\\biii\\b|\\biv\\b).*', 'i') THEN 1 ELSE 0 END) as has_roman_numeral,
    SUM(CASE WHEN NOT REGEXP_LIKE(job_title, 
        '.*(junior|entry|associate|senior|lead|principal|manager|director|\\bvp\\b|vice president).*', 'i'
    ) THEN 1 ELSE 0 END) as no_seniority_signal
FROM all_titles
GROUP BY source
"""

df = pd.read_sql(query, conn)
print("=== Pillar 2: Seniority Keyword Frequency by Source ===\n")
print(df.to_string(index=False))
print()

# Print all titles for eyeballing
query2 = """
SELECT
    'jsearch' as source,
    RAW_PAYLOAD:job_title::string as job_title
FROM RAW.JSEARCH.SRC_POSTINGS
UNION ALL
SELECT
    'theirstack' as source,
    RAW_PAYLOAD:job_title::string as job_title
FROM RAW.THEIRSTACK.SRC_POSTINGS
UNION ALL
SELECT
    'builtin' as source,
    RAW_PAYLOAD:title::string as job_title
FROM RAW.BUILTIN.SRC_POSTINGS
ORDER BY source, job_title
"""

df2 = pd.read_sql(query2, conn)
print("=== All job titles by source ===\n")
for source in ["jsearch", "theirstack", "builtin"]:
    titles = df2[df2["SOURCE"] == source]["JOB_TITLE"].tolist()
    print(f"--- {source} ({len(titles)} jobs) ---")
    for t in titles:
        print(f"  {t}")
    print()
```

```python
query = """
WITH all_jobs AS (
    SELECT
        'jsearch' as source,
        RAW_PAYLOAD:job_title::string                        as job_title,
        RAW_PAYLOAD:job_is_remote::boolean                   as is_remote,
        NULL                                                 as is_hybrid,
        RAW_PAYLOAD:job_employment_type::string              as employment_type,
        RAW_PAYLOAD:job_min_salary::float                    as min_salary,
        RAW_PAYLOAD:job_max_salary::float                    as max_salary,
        RAW_PAYLOAD:job_salary_period::string                as salary_period,
        RAW_PAYLOAD:job_location::string                     as location_string
    FROM RAW.JSEARCH.SRC_POSTINGS

    UNION ALL

    SELECT
        'theirstack' as source,
        RAW_PAYLOAD:job_title::string                        as job_title,
        RAW_PAYLOAD:remote::boolean                          as is_remote,
        RAW_PAYLOAD:hybrid::boolean                          as is_hybrid,
        RAW_PAYLOAD:employment_statuses[0]::string           as employment_type,
        RAW_PAYLOAD:min_annual_salary_usd::float             as min_salary,
        RAW_PAYLOAD:max_annual_salary_usd::float             as max_salary,
        NULL                                                 as salary_period,
        RAW_PAYLOAD:location::string                         as location_string
    FROM RAW.THEIRSTACK.SRC_POSTINGS

    UNION ALL

    SELECT
        'builtin' as source,
        RAW_PAYLOAD:title::string                            as job_title,
        CASE WHEN RAW_PAYLOAD:jobLocationType::string = 'TELECOMMUTE'
             THEN TRUE ELSE FALSE END                        as is_remote,
        NULL                                                 as is_hybrid,
        RAW_PAYLOAD:employmentType::string                   as employment_type,
        CASE WHEN RAW_PAYLOAD:baseSalary:value:unitText::string = 'YEAR'
             THEN RAW_PAYLOAD:baseSalary:value:minValue::float
             ELSE NULL END                                   as min_salary,
        CASE WHEN RAW_PAYLOAD:baseSalary:value:unitText::string = 'YEAR'
             THEN RAW_PAYLOAD:baseSalary:value:maxValue::float
             ELSE NULL END                                   as max_salary,
        RAW_PAYLOAD:baseSalary:value:unitText::string        as salary_period,
        RAW_PAYLOAD:jobLocation:address:addressLocality::string as location_string
    FROM RAW.BUILTIN.SRC_POSTINGS
)

SELECT
    source,
    COUNT(*)                                                          as total_jobs,
    -- Remote signal
    SUM(CASE WHEN is_remote = TRUE THEN 1 ELSE 0 END)                as is_remote_true,
    SUM(CASE WHEN is_remote = FALSE THEN 1 ELSE 0 END)               as is_remote_false,
    SUM(CASE WHEN is_remote IS NULL THEN 1 ELSE 0 END)               as is_remote_null,
    -- Hybrid signal (TheirStack only)
    SUM(CASE WHEN is_hybrid = TRUE THEN 1 ELSE 0 END)                as is_hybrid_true,
    SUM(CASE WHEN is_hybrid IS NULL THEN 1 ELSE 0 END)               as is_hybrid_null,
    -- Salary coverage by work model
    SUM(CASE WHEN is_remote = TRUE 
        AND min_salary IS NOT NULL THEN 1 ELSE 0 END)                as remote_with_salary,
    SUM(CASE WHEN is_remote = FALSE 
        AND min_salary IS NOT NULL THEN 1 ELSE 0 END)                as onsite_with_salary,
    -- Location string coverage
    SUM(CASE WHEN location_string IS NOT NULL 
        AND location_string != '' THEN 1 ELSE 0 END)                 as has_location_string
FROM all_jobs
GROUP BY source
"""

df = pd.read_sql(query, conn)
print("=== Pillar 3: Work Model Field Audit ===\n")
print(df.to_string(index=False))
print()

# Distribution of work model labels in location strings
# to understand what text signals we have beyond boolean fields
query2 = """
SELECT
    'jsearch' as source,
    RAW_PAYLOAD:job_title::string    as job_title,
    RAW_PAYLOAD:job_is_remote::boolean as is_remote,
    RAW_PAYLOAD:job_location::string as location_string
FROM RAW.JSEARCH.SRC_POSTINGS

UNION ALL

SELECT
    'theirstack' as source,
    RAW_PAYLOAD:job_title::string  as job_title,
    RAW_PAYLOAD:remote::boolean    as is_remote,
    RAW_PAYLOAD:location::string   as location_string
FROM RAW.THEIRSTACK.SRC_POSTINGS

UNION ALL

SELECT
    'builtin' as source,
    RAW_PAYLOAD:title::string                               as job_title,
    CASE WHEN RAW_PAYLOAD:jobLocationType::string = 'TELECOMMUTE'
         THEN TRUE ELSE FALSE END                           as is_remote,
    RAW_PAYLOAD:jobLocation:address:addressLocality::string as location_string
FROM RAW.BUILTIN.SRC_POSTINGS
ORDER BY source, is_remote
"""

df2 = pd.read_sql(query2, conn)
print("=== Work model signal per job ===\n")
for source in ["jsearch", "theirstack", "builtin"]:
    subset = df2[df2["SOURCE"] == source][["JOB_TITLE", "IS_REMOTE", "LOCATION_STRING"]]
    print(f"--- {source} ---")
    print(subset.to_string(index=False))
    print()

# Check how many jobs have "remote" or "hybrid" in their title
# as a secondary signal for JSearch which has no hybrid field
query3 = """
SELECT
    SUM(CASE WHEN REGEXP_LIKE(RAW_PAYLOAD:job_title::string, 
        '.*(remote).*', 'i') THEN 1 ELSE 0 END)            as remote_in_title,
    SUM(CASE WHEN REGEXP_LIKE(RAW_PAYLOAD:job_title::string, 
        '.*(hybrid).*', 'i') THEN 1 ELSE 0 END)            as hybrid_in_title,
    SUM(CASE WHEN REGEXP_LIKE(RAW_PAYLOAD:job_title::string, 
        '.*(on.?site|onsite|in.?office).*', 'i') 
        THEN 1 ELSE 0 END)                                  as onsite_in_title,
    COUNT(*)                                                as total
FROM RAW.JSEARCH.SRC_POSTINGS
"""

df3 = pd.read_sql(query3, conn)
print("=== JSearch: work model signal in job titles ===\n")
print(df3.to_string(index=False))
```
